```python
# Generate synthetic 2D data points
np.random.seed(42)
sample_data = np.random.randint(0, 50, size=(120, 2))

# Initialize the K-means model (e.g., 5 tries, 10 iterations, 3 clusters)
model = kmeans(n_try=10, n_iter=20, n_cluster=3)

# Fit the model to the data
final_centers = model.fit_data(sample_data)

print("\nFinal Centers Found:")
print(final_centers)
```

```python
# Plot the cluster assignments
model.plot(sample_data)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# 1. Generate synthetic data with 3 clusters
# n_samples: total number of points, centers: number of clusters (3), random_state: for reproducibility
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=42)

print("Dataset shape:", X.shape)

# 2. Initialize and fit the Scikit-Learn KMeans model
model = kmeans(n_try=5, n_iter=10, n_cluster=3)
final = model.fit_data(X)
```
